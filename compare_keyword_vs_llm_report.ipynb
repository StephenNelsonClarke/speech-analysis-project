# Keyword vs LLM theme comparison

This notebook compares the rule-based keyword themes with the LLM-assigned themes using canonical theme keys.
It automatically reads the latest saved database, LLM results, and disagreement CSV under `outputs/`.


```python
from pathlib import Path
import json
import pandas as pd

repo = Path('.')
outputs = repo / 'outputs'

def latest_file(root, filename):
    files = sorted(root.rglob(filename), key=lambda p: p.stat().st_mtime, reverse=True)
    return files[0] if files else None

database_path = latest_file(outputs, 'mpc_speech_database.csv')
llm_path = latest_file(outputs, 'llm_theme_results.jsonl')
disagreement_path = latest_file(outputs, 'theme_disagreements.csv')

db = pd.read_csv(database_path) if database_path else pd.DataFrame()
llm = pd.read_json(llm_path, lines=True) if llm_path else pd.DataFrame()
disagreements = pd.read_csv(disagreement_path) if disagreement_path else pd.DataFrame()

print(f'Database: {database_path}')
print(f'LLM results: {llm_path}')
print(f'Disagreements: {disagreement_path}')

db.head() if not db.empty else db
```

```python
THEME_LABEL_TO_KEY = {
    'Inflation & prices': 'inflation_and_prices',
    'Monetary policy & rates': 'monetary_policy_rates',
    'Growth & output': 'growth_output',
    'Labour market': 'labour_market',
    'Financial conditions': 'financial_conditions',
    'Global economy & trade': 'global_economy_trade',
    'Fiscal policy & public finances': 'fiscal_policy_public_finances',
    'Financial stability & regulation': 'financial_stability_regulation',
    'Climate & sustainability': 'climate_sustainability',
    'Uncertainty & forecasting': 'uncertainty_forecasting',
}

def normalize_theme(value):
    if value is None:
        return None
    text = str(value).strip()
    if not text:
        return None
    if text in THEME_LABEL_TO_KEY.values():
        return text
    if text in THEME_LABEL_TO_KEY:
        return THEME_LABEL_TO_KEY[text]
    cleaned = text.lower().replace(' ', '_').replace('&', 'and').replace('-', '_')
    cleaned = ''.join(ch for ch in cleaned if ch.isalnum() or ch == '_')
    return cleaned if cleaned in THEME_LABEL_TO_KEY.values() else None

def parse_theme_list(value):
    if value is None:
        return []
    if isinstance(value, list):
        return [str(v) for v in value]
    text = str(value).strip()
    if not text:
        return []
    if text.startswith('[') and text.endswith(']'):
        try:
            parsed = json.loads(text)
            if isinstance(parsed, list):
                return [str(v) for v in parsed]
        except json.JSONDecodeError:
            pass
    return [part.strip(" '\"") for part in text.split(',') if part.strip()]

def normalize_sequence(values):
    seen = set()
    normalized = []
    for value in values:
        key = normalize_theme(value)
        if key and key not in seen:
            normalized.append(key)
            seen.add(key)
    return normalized
```

```python
# Prepare canonical keys on both sides
if not db.empty:
    db['keyword_primary_key'] = db['primary_theme_label'].apply(normalize_theme)
    db['keyword_top_keys'] = db['top_themes'].apply(parse_theme_list).apply(normalize_sequence)
    db['keyword_theme_keys'] = db.apply(lambda r: normalize_sequence([r['primary_theme_label']] + parse_theme_list(r['top_themes'])), axis=1)
    db['primary_theme_label'].value_counts()
else:
    print('No database loaded.')
```

```python
if not llm.empty:
    llm['llm_primary_key'] = llm['primary_theme'].apply(normalize_theme)
    llm['llm_secondary_keys'] = llm['secondary_themes'].apply(parse_theme_list).apply(normalize_sequence)
    llm['llm_theme_keys'] = llm.apply(lambda r: normalize_sequence([r['llm_primary_key']] + r['llm_secondary_keys']), axis=1)
    llm['llm_primary_key'].value_counts()
else:
    print('No LLM results loaded.')
```

```python
# Compare canonical themes and build the same style of disagreement view used in the CSV report
if not db.empty and not llm.empty:
    merged = db.merge(
        llm[['speech_id', 'llm_primary_key', 'llm_secondary_keys', 'llm_theme_keys', 'confidence']],
        on='speech_id',
        how='left'
    )
    merged['primary_theme_agrees'] = merged['keyword_primary_key'].fillna('') == merged['llm_primary_key'].fillna('')
    merged['any_theme_overlap'] = merged.apply(lambda r: bool(set(r['keyword_theme_keys']).intersection(r['llm_theme_keys'])), axis=1)
    merged['shared_themes'] = merged.apply(lambda r: '; '.join(sorted(set(r['keyword_theme_keys']).intersection(r['llm_theme_keys']))), axis=1)
    disagreement_view = merged[(~merged['primary_theme_agrees']) | (~merged['any_theme_overlap'])].copy()
    disagreement_view[['speech_id', 'speaker', 'year', 'speech_title', 'primary_theme_label', 'llm_primary_key', 'primary_theme_agrees', 'any_theme_overlap', 'shared_themes']]
else:
    print('Cannot build comparison view because one of the inputs is missing.')
```
